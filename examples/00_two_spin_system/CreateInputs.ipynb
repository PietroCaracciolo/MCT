```python
import json
import os
import numpy as np
```

```python
def save_json(input_folder, file_name, data):
    """
    Saves a Python dictionary to a JSON file.

    Parameters:
    - input_folder (str) : The directory where the JSON file will be saved.
    - file_name (str)    : The name of the JSON file (e.g., 'params.json').
    - data (dict)        : The data to be saved in JSON format.
    """

    # If the input folder is specified and does not exist, create it
    if input_folder and not os.path.exists(input_folder):
        os.makedirs(input_folder)  # Create directory if it doesn't exist

    # Open the file in write mode and dump the data in JSON format with indentation
    with open(input_folder + file_name, "w") as file:
        json.dump(data, file, indent=2)

    # Print confirmation message with the full path of the saved file
    print(f"JSON file saved to: {input_folder}{file_name}")
```

```python
def make_batch(simulation_name, N):
    """
    Creates N simulation input folders and saves relevant parameter files for each run.

    Parameters:
    - simulation_name (str): The base folder name where all simulation input folders will be created.
    - N (int): Number of distinct simulations with the same name.
    """

    for i in range(N):
        # Define the path for this specific simulation instance
        input_folder = "./input/" + simulation_name + "/" + str(i)
        
        # Set the simulation-specific parameters
        mc_params["simulation_name"] = simulation_name
        mc_params["simulation_id"] = str(i)

        # Save model and MC parameters into JSON files
        save_json(input_folder, "/model_params.json", model_params)
        save_json(input_folder, "/mc_params.json", mc_params)

        # Optionally save the initial state and interactions if specified in model_params
        if model_params["state_initialization"] == "file":
            save_json(input_folder, "/state.json", state)
        if model_params["interactions_initialization"] == "file":
            save_json(input_folder, "/interactions.json", interactions)
```
